```python
import soundfile as sf
from IPython.display import Audio
import numpy as np

nombre_out = "output_exercise.wav"
x, fs = sf.read(nombre_out)

start_time = 30
end_time = start_time + 10

start_index = int(start_time * fs)
end_index = int(end_time * fs)

audio_segment = x[start_index:end_index, :]

Audio([audio_segment[:,1], audio_segment[:,0]],rate=fs)
```

```python
cutoff_low = 500
cutoff_high = 4000
filter_order = 8
chebyshev1_ripple = 1.0
chebyshev2_ripple = 40.0
elliptic_ripple_pass = 0.5
elliptic_ripple_stop = 50.0
bessel_norm = 'mag'
```

filtrol butterworth



```python
from scipy.signal import butter, filtfilt

num_butter, den_butter = butter(filter_order, cutoff_high, btype='lowpass', output='ba', fs=fs)
filtered_audio_butter = filtfilt(num_butter, den_butter, audio_segment, axis=0)
```

filtro chebyshev 1



```python
from scipy.signal import cheby1
from scipy.signal import filtfilt

num_cheby1, den_cheby1 = cheby1(filter_order, chebyshev1_ripple, cutoff_high, btype='lowpass', output='ba', fs=fs)
filtered_audio_cheby1 = filtfilt(num_cheby1, den_cheby1, audio_segment, axis=0)
```

filtro chebyshev 2




```python
from scipy.signal import cheby2

num_cheby2, den_cheby2 = cheby2(filter_order, chebyshev2_ripple, cutoff_high, btype='lowpass', output='ba', fs=fs)
filtered_audio_cheby2 = filtfilt(num_cheby2, den_cheby2, audio_segment, axis=0)
```

## filtro de bessel


```python
from scipy.signal import bessel, filtfilt

bessel_norm = 'mag'
num_bessel, den_bessel = bessel(filter_order, cutoff_high, btype='lowpass', norm=bessel_norm, output='ba', fs=fs)
filtered_audio_bessel = filtfilt(num_bessel, den_bessel, audio_segment, axis=0)
```

## filtro eliptico



```python
from scipy.signal import ellip, filtfilt

num_ellip, den_ellip = ellip(filter_order, elliptic_ripple_pass, elliptic_ripple_stop, cutoff_high, btype='lowpass', output='ba', fs=fs)
filtered_audio_ellip = filtfilt(num_ellip, den_ellip, audio_segment, axis=0)
```

##  resultados



```python
from IPython.display import Audio

print("Original Audio:")
display(Audio([audio_segment[:,1], audio_segment[:,0]], rate=fs))

print("Butterworth Filtered Audio:")
display(Audio([filtered_audio_butter[:,1], filtered_audio_butter[:,0]], rate=fs))

print("Chebyshev 1 Filtered Audio:")
display(Audio([filtered_audio_cheby1[:,1], filtered_audio_cheby1[:,0]], rate=fs))

print("Chebyshev 2 Filtered Audio:")
display(Audio([filtered_audio_cheby2[:,1], filtered_audio_cheby2[:,0]], rate=fs))

print("Bessel Filtered Audio:")
display(Audio([filtered_audio_bessel[:,1], filtered_audio_bessel[:,0]], rate=fs))

print("Elliptic Filtered Audio:")
display(Audio([filtered_audio_ellip[:,1], filtered_audio_ellip[:,0]], rate=fs))
```

## Explicación del método de diseño de filtros FIR por ventaneo

El método de diseño de filtros FIR por ventaneo se basa en la idea de tomar la respuesta al impulso ideal de un filtro (que es infinitamente larga) y multiplicarla por una función de ventana de longitud finita. Este proceso trunca la respuesta al impulso, lo que resulta en un filtro FIR causal y estable.

La forma de la función de ventana utilizada tiene un impacto significativo en la respuesta en frecuencia del filtro resultante. Los filtros ideales tienen transiciones abruptas en el dominio de la frecuencia, pero sus respuestas al impulso son infinitamente largas. Truncar la respuesta al impulso sin una ventana (equivalente a usar una ventana rectangular) produce un rizado considerable tanto en la banda de paso como en la banda de rechazo (fenómeno de Gibbs).

Diferentes funciones de ventana (por ejemplo, Hanning, Hamming, Blackman, Kaiser) tienen distintas características que afectan el compromiso entre el ancho de la banda de transición y la atenuación en la banda de rechazo. Ventanas más estrechas conducen a bandas de transición más anchas pero menor rizado (mejor atenuación en la banda de rechazo). Ventanas más anchas dan como resultado bandas de transición más estrechas pero mayor rizado (peor atenuación en la banda de rechazo).

La función `scipy.signal.firwin` es una herramienta común en SciPy para diseñar filtros FIR utilizando el método de ventaneo. Sus parámetros principales incluyen:

- `numtaps`: El número de coeficientes del filtro, que determina el orden del filtro. Un mayor número de coeficientes generalmente resulta en una transición más abrupta pero aumenta la carga computacional.
- `cutoff`: La(s) frecuencia(s) de corte. Para un filtro paso bajo o paso alto, es un escalar. Para un filtro paso banda o rechaza banda, es una secuencia de dos frecuencias. Las frecuencias se normalizan a la frecuencia de Nyquist (la mitad de la frecuencia de muestreo).
- `window`: La función de ventana a aplicar. Puede ser una cadena (por ejemplo, 'hamming', 'hann', 'blackman') o una tupla que especifica el tipo de ventana y sus parámetros (por ejemplo, ('kaiser', beta)). Si es `None`, se utiliza una ventana rectangular.
- `pass_zero`: Un booleano que indica si la frecuencia cero (DC) es una banda de paso. Se utiliza para especificar el tipo de filtro (paso bajo, paso alto, paso banda, rechaza banda).
- `fs`: La frecuencia de muestreo de la señal. Si se proporciona, las frecuencias de corte se interpretan en las mismas unidades que `fs`.

Ejemplos de uso de `firwin`:

- Filtro paso bajo: `firwin(numtaps, cutoff, fs=fs)` donde `cutoff` es una sola frecuencia.
- Filtro paso alto: `firwin(numtaps, cutoff, pass_zero=False, fs=fs)` donde `cutoff` es una sola frecuencia.
- Filtro paso banda: `firwin(numtaps, [low_cutoff, high_cutoff], pass_zero=False, fs=fs)` donde `[low_cutoff, high_cutoff]` es una lista de dos frecuencias.
- Filtro rechaza banda: `firwin(numtaps, [low_cutoff, high_cutoff], pass_zero=True, fs=fs)` donde `[low_cutoff, high_cutoff]` es una lista de dos frecuencias.

## Cuadro Comparativo de Filtros FIR vs IIR

| Característica        | Filtros FIR (Finite Impulse Response)                       | Filtros IIR (Infinite Impulse Response)                     |
|-----------------------|-------------------------------------------------------------|-------------------------------------------------------------|
| **Respuesta al Impulso** | Finita, se extingue después de un número finito de muestras. | Infinita, continúa indefinidamente (teóricamente).          |
| **Estabilidad**       | Siempre estables si los coeficientes son finitos.            | Pueden ser inestables si los polos están fuera del círculo unitario en el plano Z. |
| **Fase**              | Pueden tener fase lineal exacta (retardo de grupo constante). | Generalmente tienen fase no lineal.                       |
| **Complejidad**       | Requieren un mayor número de coeficientes para obtener una respuesta similar a un IIR. | Requieren menos coeficientes para obtener una respuesta similar a un FIR. |
| **Retardo**           | Introducen un retardo constante (si la fase es lineal).    | El retardo varía con la frecuencia.                       |
| **Diseño**            | Métodos comunes: ventaneo, muestreo en frecuencia, mínimos cuadrados. | Métodos comunes: Butterworth, Chebyshev, Elíptico, Bessel. |
| **Aplicaciones Típicas** | Aplicaciones donde la fase lineal es crítica (audio, procesamiento de imágenes). | Aplicaciones donde la eficiencia computacional es primordial (telecomunicaciones, control). |
| **Implementación**    | Estructura no recursiva.                                    | Estructura recursiva (requiere retroalimentación).          |
| **Fenómeno de Gibbs** | Puede ocurrir rizado en la banda de paso/rechazo con ventanas abruptas. | No presentan fenómeno de Gibbs directamente, pero el rizado depende del diseño. |
| **Requerimientos de Memoria y Cómputo** | Generalmente mayores debido al número de coeficientes. | Generalmente menores.                                       |
