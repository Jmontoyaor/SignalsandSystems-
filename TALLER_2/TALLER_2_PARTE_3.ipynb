#  Sistema en Serie con Señales Gaussianas

A continuación se desarrolla la solución para las dos configuraciones posibles de conexión en serie de los sistemas.

---

## 📌 Condiciones del problema

- **Entrada:**  
  $$ x(t) = e^{-a t^2} $$
- **Sistema A (no lineal):**  
  $$ y_A(t) = [\text{entrada}]^2 $$
- **Sistema B (LTI):**  
  Respuesta al impulso $$ h_B(t) = B e^{-b t^2} $$

---

## 🔑 Propiedad clave: Convolución de dos funciones Gaussianas

Para resolverlo, se aplica la propiedad conocida para la convolución de dos gaussianas:

$$
(g_1 * g_2)(t) = A_1 A_2 \sqrt{ \frac{\pi}{\alpha_1 + \alpha_2} } \, e^{ -\frac{ \alpha_1 \alpha_2 }{ \alpha_1 + \alpha_2 } t^2 }
$$

---

## a) Configuración: $$ x(t) \rightarrow h_B(t) \rightarrow y_A(t) \rightarrow y(t) $$

**Etapa 1: Salida del sistema B**

Se obtiene la salida de pasar la señal por el sistema LTI:

$$
z(t) = x(t) * h_B(t) = e^{-a t^2} * B e^{-b t^2}
$$

Usando la propiedad de convolución:

$$
z(t) = B \sqrt{ \frac{\pi}{a + b} } \, e^{- \frac{ab}{a + b} t^2 }
$$

**Etapa 2: Salida del sistema A**

Esta salida pasa luego por el sistema no lineal:

$$
y(t) = [z(t)]^2 = \left( B \sqrt{ \frac{\pi}{a + b} } \, e^{- \frac{ab}{a + b} t^2 } \right)^2
$$

Simplificando:

$$
y(t) = \frac{B^2 \pi}{a + b} \, e^{- \frac{2ab}{a + b} t^2 }
$$

✅ **Resultado de la configuración (a):**

$$
y(t) = \frac{B^2 \pi}{a + b} \, e^{- \frac{2ab}{a + b} t^2 }
$$

---

## b) Configuración: $$ x(t) \rightarrow y_A(t) \rightarrow h_B(t) \rightarrow y(t) $$

**Etapa 1: Salida del sistema A**

Primero se procesa la señal por el bloque no lineal:

$$
w(t) = [x(t)]^2 = [e^{-a t^2}]^2 = e^{-2a t^2}
$$

**Etapa 2: Salida del sistema B**

Luego, esta señal pasa por el sistema LTI:

$$
y(t) = w(t) * h_B(t) = e^{-2a t^2} * B e^{-b t^2}
$$

Aplicando la fórmula de convolución de gaussianas:

$$
y(t) = B \sqrt{ \frac{\pi}{2a + b} } \, e^{- \frac{2ab}{2a + b} t^2 }
$$

✅ **Resultado de la configuración (b):**

$$
y(t) = B \sqrt{ \frac{\pi}{2a + b} } \, e^{- \frac{2ab}{2a + b} t^2 }
$$

---

## 📌 Observación final

Se puede ver que los resultados difieren en cada caso:

- **Salida (a):**  
  $$ y(t) = \frac{B^2 \pi}{a + b} \, e^{- \frac{2ab}{a + b} t^2 } $$

- **Salida (b):**  
  $$ y(t) = B \sqrt{ \frac{\pi}{2a + b} } \, e^{- \frac{2ab}{2a + b} t^2 } $$

Esto confirma que **el orden de los bloques es determinante cuando se combinan sistemas lineales e invariante en el tiempo con sistemas no lineales**.  
El orden no se puede cambiar como sí sucede con bloques puramente LTI.

# Demostración de propiedades de la Transformada de Laplace




## Definición

Se parte de la definición general:

$$
X(s) = \mathcal{L}\{x(t)\} = \int_{-\infty}^{\infty} x(t) e^{-st} dt.
$$

---

## 1. Desplazamiento en el tiempo

**Demostrar:**  
$$
\mathcal{L}\{x(t - t_0)\} = e^{-s t_0} X(s).
$$

**Solución:**  
Se aplica la definición:

$$
\mathcal{L}\{x(t - t_0)\} = \int_{-\infty}^{\infty} x(t - t_0) e^{-st} dt.
$$

Cambio de variable $\tau = t - t_0$:

$$
= \int_{-\infty}^{\infty} x(\tau) e^{-s(\tau + t_0)} d\tau = e^{-s t_0} \int_{-\infty}^{\infty} x(\tau) e^{-s \tau} d\tau.
$$

Por lo tanto,

$$
\mathcal{L}\{x(t - t_0)\} = e^{-s t_0} X(s).
$$

---

## 2. Escalamiento en el tiempo

**Demostrar:**  
$$
\mathcal{L}\{x(at)\} = \frac{1}{|a|} X\!\big(\frac{s}{a}\big).
$$

**Solución:**  
A partir de la definición:

$$
\mathcal{L}\{x(at)\} = \int_{-\infty}^{\infty} x(at) e^{-st} dt.
$$

Con $\tau = at$, $dt = \frac{d\tau}{a}$:

- Si $a > 0$:

$$
= \frac{1}{a} \int_{-\infty}^{\infty} x(\tau) e^{-s \frac{\tau}{a}} d\tau = \frac{1}{|a|} X\!\big(\frac{s}{a}\big).
$$

- Si $a < 0$:

$$
\int_{\infty}^{-\infty} x(\tau) e^{-s \frac{\tau}{a}} \frac{d\tau}{a} = -\frac{1}{a} \int_{-\infty}^{\infty} x(\tau) e^{-s \frac{\tau}{a}} d\tau = \frac{1}{|a|} X\!\big(\frac{s}{a}\big).
$$

---

## 3. Derivada en el tiempo

**Demostrar:**  
$$
\mathcal{L}\left\{ \frac{dx}{dt} \right\} = s X(s).
$$

**Solución:**  
Por integración por partes:

$$
\int \frac{dx}{dt} e^{-st} dt = \Big[ x(t) e^{-st} \Big]_{-\infty}^{\infty} - \int x(t) (-s e^{-st}) dt.
$$

Asumiendo que $x(t) e^{-st}$ se anula en los extremos:

$$
= s \int_{-\infty}^{\infty} x(t) e^{-st} dt = s X(s).
$$

Para la forma unilateral:

$$
\mathcal{L}\left\{ \frac{dx}{dt} \right\} = s X(s) - x(0).
$$

---

## 4. Convolución

**Demostrar:**  
$$
\mathcal{L}\{x(t) * y(t)\} = X(s) Y(s).
$$

**Solución:**  
Definición de convolución:

$$
x(t) * y(t) = \int_{-\infty}^{\infty} x(\tau) y(t - \tau) d\tau.
$$

Entonces,

$$
\mathcal{L}\{x(t) * y(t)\} = \int_{-\infty}^{\infty} \left[ \int_{-\infty}^{\infty} x(\tau) y(t - \tau) d\tau \right] e^{-st} dt.
$$

Intercambiando integrales:

$$
= \int_{-\infty}^{\infty} x(\tau) \left[ \int_{-\infty}^{\infty} y(t - \tau) e^{-st} dt \right] d\tau.
$$

Con $\theta = t - \tau$:

$$
\int_{-\infty}^{\infty} y(t - \tau) e^{-st} dt = e^{-s\tau} Y(s).
$$

Por lo tanto,

$$
\mathcal{L}\{x(t) * y(t)\} = Y(s) \int_{-\infty}^{\infty} x(\tau) e^{-s\tau} d\tau = X(s) Y(s).
$$

# 📘 Análisis Teórico de Transformadas de Laplace

## 🎯 Objetivo

Para cada una de las siguientes señales, se busca:

1. Calcular su **Transformada de Laplace bilateral**.
2. Determinar sus **polos y ceros**.
3. Identificar la **Región de Convergencia (ROC)**.
4. Representar gráficamente el plano-$s$: polos, ceros y ROC.

---

## 🛠️ Herramientas utilizadas

- **Transformadas básicas:**

$$
\mathcal{L}\{e^{at}u(t)\} = \frac{1}{s - a}, \quad \text{ROC: Re}(s) > a
$$

$$
\mathcal{L}\{e^{at}u(-t)\} = \frac{-1}{s - a}, \quad \text{ROC: Re}(s) < a
$$

$$
\mathcal{L}\{e^{-a|t|}\} = \frac{2a}{s^2 - a^2}, \quad \text{ROC: } -a < \text{Re}(s) < a
$$

$$
\mathcal{L}\{f(t - t_0)u(t - t_0)\} = e^{-s t_0}F(s)
$$

---

## 📊 Señales analizadas

---

### 🟦 i) $$ x(t) = e^{-2t}u(t) + e^{-3t}u(t) $$

- Transformada:
  $$
  X(s) = \frac{1}{s + 2} + \frac{1}{s + 3}
  $$

- Polos: $$ s = -2,\ -3 $$
- Ceros: Ninguno
- ROC: $$ \text{Re}(s) > -2 $$

---

### 🟥 ii) $$ x(t) = e^{2t}u(t) + e^{-3t}u(-t) $$

- Transformada:
  $$
  X(s) = \frac{1}{s - 2} - \frac{1}{s + 3}
  $$

- Polos: $$ s = 2,\ -3 $$
- Ceros: Ninguno
- ROC: **No existe una región común de convergencia**  
  (la transformada bilateral **no converge**)

---

### 🟨 iii) $$ x(t) = e^{-a|t|} $$

- Equivalente a:  
  $$
  x(t) = e^{-a t}u(t) + e^{a t}u(-t)
  $$

- Transformada:
  $$
  X(s) = \frac{1}{s + a} + \frac{1}{s - a} = \frac{2a}{s^2 - a^2}
  $$

- Polos: $$ s = \pm a $$
- Ceros: Ninguno
- ROC: $$ -a < \text{Re}(s) < a $$

---

### 🟩 iv) $$ x(t) = e^{-2t}[u(t) - u(t - 5)] $$

- Señal acotada entre $$ t = 0 $$ y $$ t = 5 $$ (ventana de duración 5)

- Transformada:
  $$
  X(s) = \frac{1 - e^{-5(s + 2)}}{s + 2}
  $$

- Polos: $$ s = -2 $$
- Ceros: Infinitos (en $$ s = \frac{2\pi j n}{5} $$, con $$ n \in \mathbb{Z} $$)
- ROC: $$ \text{Re}(s) > -2 $$

---

> Este análisis se acompaña de visualizaciones gráficas en el plano-$s$ con `matplotlib`, mostrando los polos, ceros y la región de convergencia.


```python
# ✅ Análisis de Transformadas de Laplace
import numpy as np
import matplotlib.pyplot as plt
from sympy import symbols, latex, simplify, exp, Heaviside, integrate, init_printing, pprint

init_printing()

# Señales y parámetros
signals = [
    {
        "name": "i) $e^{-2t}u(t) + e^{-3t}u(t)$",
        "transform": r"X(s) = \frac{1}{s + 2} + \frac{1}{s + 3}",
        "poles": [-2, -3],
        "zeros": [],
        "roc_type": "right",
        "roc_val": -2
    },
    {
        "name": "ii) $e^{2t}u(t) + e^{-3t}u(-t)$",
        "transform": r"X(s) = \frac{1}{s - 2} - \frac{1}{s + 3}",
        "poles": [2, -3],
        "zeros": [],
        "roc_type": "none",
    },
    {
        "name": "iii) $e^{-a|t|}$",
        "transform": r"X(s) = \frac{2a}{s^2 - a^2}",
        "poles": None,  # handled with a
        "zeros": [],
        "roc_type": "band",
        "a": 1.0
    },
    {
        "name": "iv) $e^{-2t}[u(t) - u(t - 5)]$",
        "transform": r"X(s) = \frac{1 - e^{-5(s + 2)}}{s + 2}",
        "poles": [-2],
        "zeros": [2 * np.pi * 1j * n / 5 for n in range(-2, 3)],  # muestras de ceros
        "roc_type": "right",
        "roc_val": -2
    }
]

# Función para graficar polos, ceros y ROC
def plot_pz_roc(poles, zeros, roc_type, roc_val=None, a_val=None, title="Plano s"):
    fig, ax = plt.subplots(figsize=(6, 4))

    # Ceros y polos
    if zeros:
        ax.scatter([z.real for z in zeros], [z.imag for z in zeros], marker='o', color='blue', label='Ceros')
    if poles:
        ax.scatter(poles, [0]*len(poles), marker='x', color='red', label='Polos')

    # ROC
    x = np.linspace(-5, 5, 400)
    y = np.linspace(-3, 3, 200)
    X, Y = np.meshgrid(x, y)
    if roc_type == 'right':
        mask = X > roc_val
        ax.contourf(X, Y, mask, levels=[0.5, 1], colors=['green'], alpha=0.2)
    elif roc_type == 'band' and a_val is not None:
        mask = (X > -a_val) & (X < a_val)
        ax.contourf(X, Y, mask, levels=[0.5, 1], colors=['green'], alpha=0.2)

    # Ejes
    ax.axhline(0, color='black', linewidth=0.5)
    ax.axvline(0, color='black', linewidth=0.5)
    ax.set_xlabel('Re{s}')
    ax.set_ylabel('Im{s}')
    ax.set_title(title)
    ax.legend(loc='upper right')
    ax.set_aspect('equal', 'box')
    plt.grid(True)
    plt.show()

# Recorremos cada señal y mostramos resultados
for sig in signals:
    print("====================================================")
    print(f"🔹 Análisis de la señal: {sig['name']}")
    print("----------------------------------------------------")
    print("📌 Transformada de Laplace:")
    print(sig["transform"])

    # Polos y ceros
    if sig.get("name", "").startswith("iii"):
        a = sig["a"]
        poles = [-a, a]
    else:
        poles = sig.get("poles", [])
    zeros = sig.get("zeros", [])

    print(f"📍 Polos: {poles if poles else 'Ninguno finito'}")
    print(f"🌀 Ceros: {zeros if zeros else 'Ninguno finito'}")

    # ROC
    print("📈 Región de Convergencia (ROC):")
    if sig["roc_type"] == "right":
        print(f"   Re(s) > {sig['roc_val']}")
    elif sig["roc_type"] == "band":
        print(f"   -a < Re(s) < a, con a = {sig['a']}")
    else:
        print("   ROC vacía (no converge bilateralmente)")

    # Gráfico
    plot_pz_roc(poles, zeros, sig["roc_type"], sig.get("roc_val"), sig.get("a"), title=sig["name"])
```

## ⚙️ Funciones de Transferencia en Lazo Abierto

---

### 🔷 Caso 1: Circuito RLC Serie

- Entrada: \( V_{\text{in}}(s) \)
- Salida: \( V_C(s) \)

Aplicando la ley de voltajes:

$$
V_{\text{in}}(s) = \left( R + Ls + \frac{1}{C s} \right) I(s)
$$

Y usando \( V_C(s) = \frac{1}{C s} I(s) \), se obtiene:

$$
H_{\text{serie}}(s) = \frac{V_C(s)}{V_{\text{in}}(s)} = \frac{1}{LC s^2 + RC s + 1}
$$

---

### 🔷 Caso 2: Circuito RLC Paralelo

- Entrada: \( I_{\text{in}}(s) \)
- Salida: \( I_L(s) \)

Aplicando ley de corrientes:

$$
Y_{\text{eq}}(s) = \frac{1}{R} + sC + \frac{1}{sL}
$$

Entonces:

$$
I_L(s) = \frac{1}{sL} \cdot \frac{I_{\text{in}}(s)}{Y_{\text{eq}}(s)}
$$

Y la función de transferencia:

$$
H_{\text{paralelo}}(s) = \frac{I_L(s)}{I_{\text{in}}(s)} = \frac{1}{LC s^2 + \frac{L}{R}s + 1}
$$

---

✅ Ambas funciones de transferencia fueron obtenidas considerando **condiciones iniciales cero**, como se asume en análisis de redes con Laplace.


```python
import sympy as sp
from IPython.display import display

# Variables simbólicas
s, R, L, C = sp.symbols('s R L C')

# =============================
# 🔷 Circuito RLC en SERIE
# =============================

# Impedancia total en serie
Z_total_serie = R + L*s + 1/(C*s)

# Función de transferencia: Vc(s)/Vin(s)
H_serie = (1/(C*s)) / Z_total_serie
H_serie_simplified = sp.simplify(H_serie)

# =============================
# 🔷 Circuito RLC en PARALELO
# =============================

# Admitancia total en paralelo
Y_total_paralelo = 1/R + s/C + 1/(s*L)

# Función de transferencia: IL(s)/Iin(s)
H_paralelo = (1/(s*L)) / Y_total_paralelo
H_paralelo_simplified = sp.simplify(H_paralelo)

# =============================
# 📤 Mostrar resultados
# =============================

print("✅ Función de transferencia RLC SERIE (salida: tensión en el capacitor):")
display(H_serie_simplified)

print("\n✅ Función de transferencia RLC PARALELO (salida: corriente en el inductor):")
display(H_paralelo_simplified)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Valores numéricos
R_val = 1.0      # ohmios
L_val = 1.0e-3   # henrios
C_val = 1.0e-6   # faradios

# Rango de frecuencia (logarítmico)
omega = np.logspace(2, 6, 1000)  # 100 Hz a 1 MHz
s = 1j * omega  # s = jω

# =============================
# H(s) para RLC SERIE
# =============================
H_serie = 1 / (L_val*C_val*s**2 + R_val*C_val*s + 1)

# =============================
# H(s) para RLC PARALELO
# =============================
H_paralelo = 1 / (L_val*C_val*s**2 + (L_val/R_val)*s + 1)

# =============================
# Plot: Magnitud y Fase
# =============================
fig, axs = plt.subplots(2, 2, figsize=(12, 8))

# --- Magnitud Serie
axs[0, 0].semilogx(omega, 20*np.log10(np.abs(H_serie)))
axs[0, 0].set_title('Magnitud |H_serie(jω)|')
axs[0, 0].set_ylabel('Magnitud (dB)')
axs[0, 0].grid(True, which='both')

# --- Fase Serie
axs[1, 0].semilogx(omega, np.angle(H_serie, deg=True))
axs[1, 0].set_title('Fase ∠H_serie(jω)')
axs[1, 0].set_ylabel('Fase (°)')
axs[1, 0].set_xlabel('Frecuencia (rad/s)')
axs[1, 0].grid(True, which='both')

# --- Magnitud Paralelo
axs[0, 1].semilogx(omega, 20*np.log10(np.abs(H_paralelo)))
axs[0, 1].set_title('Magnitud |H_paralelo(jω)|')
axs[0, 1].set_ylabel('Magnitud (dB)')
axs[0, 1].grid(True, which='both')

# --- Fase Paralelo
axs[1, 1].semilogx(omega, np.angle(H_paralelo, deg=True))
axs[1, 1].set_title('Fase ∠H_paralelo(jω)')
axs[1, 1].set_ylabel('Fase (°)')
axs[1, 1].set_xlabel('Frecuencia (rad/s)')
axs[1, 1].grid(True, which='both')

plt.tight_layout()
plt.show()
```

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------------
# 1. Símbolos
s, t, R, L, C = sp.symbols('s t R L C', real=True, positive=True)

# -----------------------------------
# 2. Función de transferencia en lazo abierto (RLC serie)
H_open = 1 / (L*C*s**2 + R*C*s + 1)

# -----------------------------------
# 3. Función de transferencia en lazo cerrado (realimentación unitaria negativa)
H_closed = H_open / (1 + H_open)
H_closed = sp.simplify(H_closed)

# -----------------------------------
# 4. Entradas en Laplace
U_impulse = 1                         # δ(t) ↔ 1
U_step = 1/s                         # u(t) ↔ 1/s
U_ramp = 1/s**2                      # rampa ↔ 1/s²

# -----------------------------------
# 5. Salidas en Laplace
Y_impulse = H_closed * U_impulse
Y_step = H_closed * U_step
Y_ramp = H_closed * U_ramp

# -----------------------------------
# 6. Transformada inversa simbólica
y_impulse_t = sp.inverse_laplace_transform(Y_impulse, s, t)
y_step_t = sp.inverse_laplace_transform(Y_step, s, t)
y_ramp_t = sp.inverse_laplace_transform(Y_ramp, s, t)

# -----------------------------------
# 7. Mostrar resultados simbólicos
sp.init_printing()
print("🔁 Lazo cerrado:")
display(H_closed)

print("\n📤 Salida ante impulso:")
display(y_impulse_t)

print("\n📤 Salida ante escalón:")
display(y_step_t)

print("\n📤 Salida ante rampa:")
display(y_ramp_t)
```

```python
# Parámetros numéricos
R_val = 1.0     # ohm
L_val = 1e-3    # H
C_val = 1e-6    # F

# Crear funciones numéricas
f_imp = sp.lambdify(t, y_impulse_t.subs({R: R_val, L: L_val, C: C_val}), 'numpy')
f_step = sp.lambdify(t, y_step_t.subs({R: R_val, L: L_val, C: C_val}), 'numpy')
f_ramp = sp.lambdify(t, y_ramp_t.subs({R: R_val, L: L_val, C: C_val}), 'numpy')

# Tiempo
tt = np.linspace(0, 0.01, 1000)

# Graficar
plt.figure(figsize=(12, 6))
plt.plot(tt, f_imp(tt), label='Impulso')
plt.plot(tt, f_step(tt), label='Escalón')
plt.plot(tt, f_ramp(tt), label='Rampa')
plt.title("Salida del sistema RLC serie en lazo cerrado")
plt.xlabel("Tiempo (s)")
plt.ylabel("y(t)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# 🔁 Respuesta temporal generalizada para sistemas RLC en lazo cerrado (serie y paralelo)
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# ------------------------
# 1. Símbolos simbólicos
s, t = sp.symbols('s t', real=True)
R, L, C = sp.symbols('R L C', positive=True)

# ------------------------
# 2. Funciones de transferencia en lazo abierto
H_serie_open = (1 / (C*s)) / (R + L*s + 1/(C*s))
H_paralelo_open = (1 / (s*L)) / (1/R + s/C + 1/(s*L))

# ------------------------
# 3. Funciones en lazo cerrado
H_serie_closed = sp.simplify(H_serie_open / (1 + H_serie_open))
H_paralelo_closed = sp.simplify(H_paralelo_open / (1 + H_paralelo_open))

# ------------------------
# 4. Entradas en Laplace
entradas = {
    "Impulso": 1,
    "Escalón": 1/s,
    "Rampa": 1/s**2
}

# ------------------------
# 5. Configuraciones
sistemas = {
    "RLC serie": H_serie_closed,
    "RLC paralelo": H_paralelo_closed
}

# ------------------------
# 6. Parámetros numéricos
R_val = 1.0      # ohm
L_val = 1e-3     # henrio
C_val = 1e-6     # faradio

# ------------------------
# 7. Tiempo y figura
tt = np.linspace(0, 0.01, 1000)
plt.figure(figsize=(12, 8))

# ------------------------
# 8. Evaluación general
sp.init_printing()
for i, (sist_nombre, Hc) in enumerate(sistemas.items(), 1):
    for j, (entrada_nombre, Xs) in enumerate(entradas.items(), 1):
        Ys = sp.simplify(Hc * Xs)
        yt = sp.inverse_laplace_transform(Ys, s, t)

        # Sustituir valores numéricos
        yt_num = yt.subs({R: R_val, L: L_val, C: C_val})
        f_yt = sp.lambdify(t, yt_num, 'numpy')
        yt_vals = f_yt(tt)

        # Graficar
        plt.subplot(2, 3, (i - 1)*3 + j)
        plt.plot(tt, yt_vals, label=f'{entrada_nombre}')
        plt.title(f'{sist_nombre} - Entrada: {entrada_nombre}')
        plt.xlabel('Tiempo (s)')
        plt.ylabel('y(t)')
        plt.grid(True)
        plt.legend()

plt.suptitle("Respuestas en lazo cerrado para RLC serie y paralelo", fontsize=14)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

```python
%%writefile 0_👋_Hello.py
import streamlit as st
import numpy as np
import matplotlib.pyplot as plt
import scipy.signal as signal

# ✅ Estilo visual moderno
st.markdown("""
    <style>
    body {
        background: linear-gradient(to right, #e3f2fd, #e0f7fa);
    }
    .stApp {
        background-color: #f5f5f5;
        font-family: 'Segoe UI', sans-serif;
        padding: 1rem;
    }
    h1, h2, h3 {
        color: #1565c0;
    }
    .block-container {
        padding-top: 1rem;
    }
    .stMarkdown > div {
        background-color: #ffffffdd;
        border-radius: 8px;
        padding: 1rem;
        box-shadow: 0 4px 8px rgba(0,0,0,0.1);
    }
    </style>
""", unsafe_allow_html=True)

# 🧭 Configuración general
st.set_page_config(layout="wide", page_title="SLIT RLC", page_icon="📉")
st.title("🔬 Análisis Dinámico de Sistemas RLC en Segundo Orden")

# 🧾 Introducción
st.markdown(r"""
Este simulador interactivo permite analizar sistemas de segundo orden basados en circuitos **RLC serie y paralelo** bajo diferentes condiciones de amortiguamiento:

- **Subamortiguado** $(0 < \zeta < 1)$
- **Críticamente amortiguado** $(\zeta = 1)$
- **Sobreamortiguado** $(\zeta > 1)$
- **Inestable** $(\zeta < 0)$

Se calcula la **frecuencia natural** $(\omega_n)$ y el **factor de amortiguamiento** $(\zeta)$ en función de los parámetros $R$, $L$ y $C$.

### Salidas disponibles:
- 📈 **Respuestas temporales** a impulso, escalón y rampa
- 🧮 **Diagrama de polos y ceros**
- 📉 **Diagrama de Bode** (magnitud y fase)
- 📊 **Métricas de rendimiento**:
  - Tiempo de subida
  - Tiempo de pico
  - Máximo sobreimpulso
  - Tiempo de establecimiento
""")

# 🎛️ Simulación
st.header("🎛️ Simulación Interactiva")

# 📌 Sidebar interactiva
st.sidebar.title("🔧 Parámetros del Sistema")
circuito = st.sidebar.selectbox("Tipo de circuito", ["Serie", "Paralelo"])
tipo = st.sidebar.selectbox("Tipo de sistema", ["Subamortiguado", "Crítico", "Sobreamortiguado", "Inestable"])

# Configuración de zeta
if tipo == "Subamortiguado":
    zeta = st.sidebar.slider("ζ", 0.01, 0.99, 0.5)
elif tipo == "Crítico":
    zeta = 1.0
elif tipo == "Sobreamortiguado":
    zeta = st.sidebar.slider("ζ", 1.01, 5.0, 2.0)
else:
    zeta = st.sidebar.slider("ζ (negativo)", -2.0, -0.01, -0.5)

wn = st.sidebar.slider("ωₙ (rad/s)", 0.1, 50.0, 10.0)

# 📘 Información adicional
with st.sidebar.expander("ℹ️ Sobre este Proyecto"):
    st.write(
        """
        - **Autor:** JUAN FERNANDO MONTOYA ORTIZ
        - **Asignatura:** Señales y Sistemas
        - **Docente:** Andres Marino Alvarez
        - **Tecnologías:**
            - `Streamlit`
            - `Python`
            - `Matplotlib`
            - `Numpy`
            - `Scikit-learn`, `yt-dlp`, `FFmpeg`
        """
    )

# 📐 Cálculo de parámetros
if circuito == "Serie":
    L = 1.0
    C = 1.0 / (wn**2 * L)
    R = 2 * zeta * wn * L
else:
    C = 1.0
    L = 1.0 / (wn**2 * C)
    R = 1.0 / (2 * zeta * wn * C)

# ℹ️ Mostrar parámetros
st.markdown(f"""
### ⚙️ Parámetros del circuito `{circuito.upper()}`
- $R$ = {R:.3f} Ω
- $L$ = {L:.3f} H
- $C$ = {C:.3f} F
""")

# 📈 Transferencia y simulaciones
num = [wn**2]
den = [1, 2*zeta*wn, wn**2]
sys = signal.TransferFunction(num, den)

t = np.linspace(0, 5, 1000)
t_imp, y_imp = signal.impulse(sys, T=t)
t_step, y_step = signal.step(sys, T=t)
_, y_ramp, _ = signal.lsim(sys, U=t, T=t)

# 🔍 Métricas
Mp = (np.max(y_step)-1)*100 if zeta < 1 else 0.0
tp = t_step[np.argmax(y_step)] if zeta < 1 else np.nan
tr = t_step[np.where(y_step >= 0.9)[0][0]] if np.any(y_step >= 0.9) else np.nan
ts = t_step[np.where(np.abs(y_step-1) <= 0.02)[0][0]] if np.any(np.abs(y_step-1) <= 0.02) else np.nan

# 🎨 Visualización
col1, col2 = st.columns(2)

with col1:
    st.subheader("📈 Impulso y Escalón")
    fig1, ax = plt.subplots(2, 1, figsize=(6, 6))
    ax[0].plot(t_imp, y_imp, color='#ff6f61')
    ax[0].set_title("Respuesta al Impulso")
    ax[0].grid(True)
    ax[1].plot(t_step, y_step, color='#29b6f6')
    ax[1].set_title("Respuesta al Escalón")
    ax[1].grid(True)
    st.pyplot(fig1)

with col2:
    st.subheader("📈 Rampa y Polos/Ceros")
    fig2, ax = plt.subplots()
    ax.plot(t, y_ramp, color='#ab47bc')
    ax.set_title("Respuesta a Rampa")
    ax.set_xlabel("t (s)")
    ax.set_ylabel("y(t)")
    ax.grid(True)
    st.pyplot(fig2)

    fig3, ax = plt.subplots()
    ax.plot(np.real(sys.poles), np.imag(sys.poles), 'x', label="Polos", color='red')
    ax.plot(np.real(sys.zeros), np.imag(sys.zeros), 'o', label="Ceros", color='blue')
    ax.axhline(0, color='gray')
    ax.axvline(0, color='gray')
    ax.set_title("Diagrama de Polos y Ceros")
    ax.set_xlabel("Re(s)")
    ax.set_ylabel("Im(s)")
    ax.legend()
    ax.grid(True)
    st.pyplot(fig3)

# Bode
st.subheader("📉 Diagrama de Bode")
fig_bode, (ax2, ax3) = plt.subplots(2, 1, figsize=(7, 5))
w, mag, phase = signal.bode(sys)
ax2.semilogx(w, mag, color='green')
ax2.set_title("Magnitud (dB)")
ax2.grid(True)
ax3.semilogx(w, phase, color='purple')
ax3.set_title("Fase (°)")
ax3.grid(True)
st.pyplot(fig_bode)

# Métricas de rendimiento
st.subheader("📊 Métricas de Rendimiento")
st.markdown(f"""
- **Sobreimpulso (Mp):** {Mp:.2f} %
- **Tiempo de Pico (tp):** {tp:.3f} s
- **Tiempo de Subida (tr):** {tr:.3f} s
- **Tiempo de Establecimiento (ts):** {ts:.3f} s
""")
```
