Aquí tienes una explicación de las similitudes y diferencias entre la Serie de Fourier, las Transformadas de Fourier (Continua y Discreta), la Transformada de Laplace y la Transformada Z:

**Similitudes:**

*   **Todas transforman señales a un dominio diferente:** La idea central detrás de todas estas transformadas es representar una señal (que generalmente se encuentra en el dominio del tiempo o espacial) en un dominio diferente, a menudo el dominio de la frecuencia o de la frecuencia compleja. Este cambio de dominio puede simplificar el análisis y la manipulación de la señal.
*   **Se utilizan para el análisis de sistemas:** Todas son herramientas poderosas para analizar el comportamiento de sistemas lineales invariantes en el tiempo (LTI). Al transformar la entrada del sistema y la respuesta al impulso al dominio transformado, la convolución se convierte en multiplicación, lo cual es mucho más sencillo de manejar.
*   **Se basan en funciones exponenciales:** Cada transformada utiliza funciones exponenciales complejas como base para la transformación. Estas funciones tienen propiedades especiales que las hacen adecuadas para representar señales en diferentes dominios.

**Diferencias:**

| Característica      | Serie de Fourier                                   | Transformada de Fourier en Tiempo Continuo (CTFT) | Transformada de Fourier en Tiempo Discreto (DTFT) | Transformada de Laplace (LT)                      | Transformada Z (ZT)                            |
| :------------------ | :------------------------------------------------- | :---------------------------------------------- | :---------------------------------------------- | :------------------------------------------ | :------------------------------------------- |
| **Tipo de Señal**   | Periódica, tiempo continuo                         | Aperiódica, tiempo continuo                     | Aperiódica, tiempo discreto                     | Tiempo continuo (permite señales inestables) | Tiempo discreto (permite señales inestables) |
| **Dominio**         | Frecuencia discreta (armónicos)                    | Frecuencia continua                             | Frecuencia continua                             | Frecuencia compleja (s = σ + jω)             | Frecuencia compleja (z = r * e^(jω))         |
| **Invertibilidad**  | Sí (reconstruye la señal periódica)                 | Sí (reconstruye la señal de tiempo continuo)       | Sí (reconstruye la señal de tiempo discreto)      | Sí (si se conoce la Región de Convergencia) | Sí (si se conoce la Región de Convergencia) |
| **Aplicabilidad**   | Representar señales periódicas como suma de sinusoides | Analizar el contenido de frecuencia de señales continuas aperiódicas | Analizar el contenido de frecuencia de señales discretas aperiódicas | Analizar sistemas LTI con condiciones iniciales y sistemas inestables | Analizar sistemas LTI con condiciones iniciales y sistemas inestables |
| **Relación**        | Puede verse como un caso especial de la CTFT para señales periódicas | Generalización de la Serie de Fourier a señales aperiódicas | Análoga a la CTFT para señales de tiempo discreto | Generalización de la CTFT al plano s complejo | Análoga a la Transformada de Laplace para señales de tiempo discreto |
| **Convergencia**    | Converge si la señal satisface las condiciones de Dirichlet | Puede no converger para todas las señales (ej. sistemas inestables) | Converge si la señal es absolutamente sumable | La convergencia depende de la Región de Convergencia (ROC) | La convergencia depende de la Región de Convergencia (ROC) |

**Puntos Clave:**

*   La **Serie de Fourier** es para señales continuas periódicas, representándolas como una suma de componentes de frecuencia discretas.
*   Las **Transformadas de Fourier (CTFT y DTFT)** extienden esta idea a señales aperiódicas en tiempo continuo y discreto, respectivamente, proporcionando una representación de frecuencia continua.
*   La **Transformada de Laplace** y la **Transformada Z** son generalizaciones de las Transformadas de Fourier al plano complejo. Son más potentes porque pueden manejar señales inestables y sistemas con condiciones iniciales, lo cual es crucial para analizar el comportamiento transitorio de los sistemas.
*   La **Región de Convergencia (ROC)** es un concepto crítico para las Transformadas de Laplace y Z, ya que determina la unicidad de la transformada y la estabilidad del sistema.

Estas transformadas son herramientas fundamentales en el procesamiento de señales, comunicaciones y sistemas de control, proporcionando diferentes perspectivas y capacidades para analizar y manipular señales y sistemas.

#Encontrar la transformada Z de las siguientes señales

Se utiliza la definición de la **Transformada Z**:
$$
X(z) = \sum_{n=-\infty}^{\infty} x[n] z^{-n}
$$

---

##i) $x[n] = -a^n u[-n - 1]$


La función $u[-n - 1] = 1$ para $n \leq -1$ y 0 en otro caso.

Entonces:
$$
x[n] = \begin{cases}
-a^n, & n \leq -1 \\
0, & \text{en otro caso}
\end{cases}
$$

Sustituyendo en la definición:
$$
X(z) = \sum_{n=-\infty}^{-1} (-a^n) z^{-n} = -\sum_{n=-\infty}^{-1} (a z)^n
$$

Esta es una **serie geométrica** hacia $-\infty$, por lo que hacemos un cambio de variable: sea $m = -n$, entonces $n = -m$ y cuando $n \to -\infty$, $m \to \infty$:
$$
X(z) = -\sum_{m=1}^{\infty} (a z)^{-m} = -\sum_{m=1}^{\infty} \left( \frac{1}{a z} \right)^m
$$

Esto es una serie geométrica con razón $r = \frac{1}{a z}$, válida cuando $|a z| > 1$, o sea:
- **ROC:** $|z| > \frac{1}{|a|}$

Entonces:
$$
X(z) = -\frac{\frac{1}{a z}}{1 - \frac{1}{a z}} = -\frac{1}{a z - 1}
$$

---

##ii) $x[n] = a^n$ para $0 \leq n \leq N - 1$, con $a > 0$


Esta es una señal **finita** de $n = 0$ a $n = N - 1$.

Aplicamos la definición:
$$
X(z) = \sum_{n=0}^{N-1} a^n z^{-n} = \sum_{n=0}^{N-1} (a z^{-1})^n
$$

Es una **serie geométrica finita** de $N$ términos:
$$
X(z) = \frac{1 - (a z^{-1})^N}{1 - a z^{-1}}
$$

O también, multiplicando numerador y denominador por $z^N$:
$$
X(z) = \frac{z^N - a^N}{z^{N-1}(z - a)}
$$

- **ROC:** Todo el plano $z$, excepto posiblemente $z = a$ si se toma como polo.

---

##iii) $x[n] = \{5, 3, -2, 0, 4, -3\}$ con $x[0] = 5$


Esta es una **secuencia finita** definida entre $n = 0$ y $n = 5$.

Aplicamos directamente:
$$
X(z) = 5 z^{0} + 3 z^{-1} -2 z^{-2} + 0 z^{-3} + 4 z^{-4} -3 z^{-5}
$$

Entonces:
$$
X(z) = 5 + 3z^{-1} - 2z^{-2} + 4z^{-4} - 3z^{-5}
$$

- **ROC:** Todo el plano $z$, excepto $z = 0$ (por los términos negativos en potencia).

## Filtros de Respuesta Impulso Finita (FIR)

Un filtro de respuesta impulso finita (FIR) es un tipo de filtro digital cuya respuesta al impulso es de duración finita, lo que significa que vuelve a cero en un número finito de pasos de tiempo. Esto contrasta con los filtros de respuesta impulso infinita (IIR), cuya respuesta al impulso teóricamente dura indefinidamente.

**Características principales de los filtros FIR:**

*   **Respuesta al Impulso Finita:** La característica definitoria de un filtro FIR es que su respuesta a una señal de impulso unitario dura un número finito de muestras. Esto se debe a que la salida del filtro es una suma ponderada de las muestras de entrada actuales y pasadas, sin depender de las salidas pasadas (no hay realimentación).
*   **Fase Lineal:** Los filtros FIR se pueden diseñar fácilmente para tener una fase exactamente lineal. Esto significa que todas las componentes de frecuencia de la señal de entrada experimentan el mismo retardo de tiempo al pasar por el filtro, lo cual es crucial en aplicaciones donde la distorsión de fase debe ser mínima (por ejemplo, en audio y comunicaciones).
*   **Estabilidad Inherente:** Todos los filtros FIR son inherentemente estables. La estabilidad de un filtro FIR está garantizada siempre que los coeficientes del filtro sean finitos, lo cual siempre es el caso en la práctica. Esto se debe a la ausencia de polos en el plano z fuera del círculo unitario.

**Ventajas de los filtros FIR:**

*   **Fácil Diseño de Fase Lineal:** Como se mencionó, es relativamente sencillo diseñar filtros FIR con una fase lineal exacta.
*   **Estabilidad Garantizada:** La estabilidad es inherente y no requiere una consideración especial durante el diseño, a diferencia de los filtros IIR.
*   **Ausencia de Realimentación:** La estructura sin realimentación simplifica su implementación y análisis, y evita problemas potenciales asociados con la realimentación, como la inestabilidad y los ciclos límite.

**Desventajas de los filtros FIR:**

*   **Mayor Orden:** Para lograr una especificación de respuesta en frecuencia dada (por ejemplo, un cierto nivel de atenuación en la banda de parada), un filtro FIR generalmente requiere un orden mucho mayor (y por lo tanto más coeficientes) que un filtro IIR equivalente.
*   **Mayor Costo Computacional:** Un orden mayor implica más multiplicaciones y sumas por cada muestra de salida, lo que resulta en un mayor costo computacional (más operaciones por segundo) y un mayor retardo de procesamiento en comparación con un filtro IIR con especificaciones similares.

## Comparación entre Invarianza al Impulso y Transformación Bilineal

**Invarianza al Impulso (II)**  
Convierte una respuesta impulsiva continua en una discreta muestreando la respuesta impulsiva del sistema analógico. Conserva la forma de la respuesta impulsiva original, pero no preserva la correspondencia exacta de la frecuencia, lo que puede producir aliasing.

**Transformación Bilineal (TB)**  
Transforma el dominio \( s \) del sistema analógico al dominio \( z \) mediante una transformación no lineal. Evita el aliasing, ya que comprime toda la frecuencia analógica en la banda discreta \([0, \pi]\). Sin embargo, introduce distorsión de frecuencia (warping), que se puede corregir mediante pre-warping.

### Comparación

| Invarianza al Impulso | Transformación Bilineal |
|-----------------------|-------------------------|
| Preserva la respuesta impulsiva | No preserva la respuesta impulsiva directamente |
| Puede producir aliasing | Evita el aliasing |
| No hay distorsión de frecuencia, pero sí aliasing | Hay distorsión de frecuencia (warping) |
| Adecuada para filtros pasa-bajos de banda estrecha | Adecuada para una amplia gama de filtros |
| No existe correspondencia exacta entre bandas de frecuencia analógicas y discretas | Relación uno a uno entre frecuencias analógicas y digitales (con warping) |

```python
%%writefile Presentation.py
import streamlit as st
import numpy as np
from scipy import signal
import matplotlib.pyplot as plt

# --- Funciones para el Diseño y Visualización de Filtros ---

def plot_bode(w, mag, phase, filter_type):
    """Genera el diagrama de Bode."""
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 6))
    fig.suptitle(f'Diagrama de Bode - Filtro {filter_type}')

    # Magnitud
    ax1.semilogx(w, mag)
    ax1.set_ylabel('Magnitud [dB]')
    ax1.grid(which='both', axis='both')
    ax1.set_title('Respuesta en Magnitud')

    # Fase
    ax2.semilogx(w, phase)
    ax2.set_xlabel('Frecuencia [rad/muestra]')
    ax2.set_ylabel('Fase [grados]')
    ax2.grid(which='both', axis='both')
    ax2.set_title('Respuesta en Fase')

    plt.tight_layout()
    st.pyplot(fig)

def plot_pole_zero(z, p, k, filter_type):
    """Genera el diagrama de polos y ceros."""
    fig, ax = plt.subplots(figsize=(6, 6))
    ax.set_title(f'Diagrama de Polos y Ceros - Filtro {filter_type}')

    # Dibuja el círculo unitario
    uc = plt.Circle((0, 0), 1, fill=False, color='gray', ls='--')
    ax.add_artist(uc)

    # Dibuja los polos y ceros
    ax.scatter(np.real(z), np.imag(z), marker='o', s=100, facecolors='none', edgecolors='b', label='Ceros')
    ax.scatter(np.real(p), np.imag(p), marker='x', s=100, color='r', label='Polos')

    ax.set_xlabel('Parte Real')
    ax.set_ylabel('Parte Imaginaria')
    ax.grid(True)
    ax.legend()
    ax.set_aspect('equal')
    st.pyplot(fig)


# --- Interfaz de Usuario de Streamlit ---

st.set_page_config(page_title="Diseño de Filtros para Demodulación AM", layout="wide")
st.title("Diseño de Filtro Pasabajas para Demodulación AM 📡")

st.markdown("""
Esta aplicación permite diseñar y visualizar filtros pasabajas que son cruciales en la demodulación de señales AM para recuperar la señal de información. Puedes escoger entre filtros **FIR** y **IIR** y ajustar sus parámetros.
""")

# --- Parámetros de Entrada del Usuario ---
st.sidebar.header("Configuración del Filtro")
fs = 48000  # Frecuencia de muestreo (ejemplo)
cutoff_freq = 4000  # Frecuencia de corte (ejemplo para señal de voz)

st.sidebar.info(f"Frecuencia de Muestreo: **{fs/1000} kHz**")
st.sidebar.info(f"Frecuencia de Corte: **{cutoff_freq/1000} kHz**")

filter_type_option = st.sidebar.selectbox("Tipo de Filtro", ["IIR", "FIR"])

if filter_type_option == "FIR":
    st.sidebar.subheader("Parámetros del Filtro FIR")
    fir_order = st.sidebar.slider("Orden del Filtro", 2, 100, 30)
    fir_window = st.sidebar.selectbox("Ventana", ["hamming", "hann", "blackman", "bartlett"])

    # Diseño del filtro FIR
    numtaps = fir_order + 1
    b = signal.firwin(numtaps, cutoff_freq, window=fir_window, fs=fs, pass_zero='lowpass')
    a = [1.0] # El denominador de un filtro FIR es 1
    z, p, k = signal.tf2zpk(b, a)
    w, h = signal.freqz(b, a, worN=2048, fs=fs)
    w_rad = w * 2 * np.pi / fs # Frecuencia en rad/muestra para el gráfico de Bode

    # Visualización
    st.header(f"Filtro FIR (Ventana {fir_window.capitalize()})")
    plot_bode(w_rad, 20 * np.log10(abs(h)), np.angle(h, deg=True), "FIR")
    st.header("Diagrama de Polos y Ceros")
    plot_pole_zero(z, p, k, "FIR")


elif filter_type_option == "IIR":
    st.sidebar.subheader("Parámetros del Filtro IIR")
    iir_arch = st.sidebar.selectbox("Arquitectura IIR", ["Butterworth", "Chebyshev", "Bessel"])
    iir_order = st.sidebar.slider("Orden del Filtro", 1, 12, 4)

    # Normalización de la frecuencia de corte
    nyquist = 0.5 * fs
    normal_cutoff = cutoff_freq / nyquist

    # Parámetros adicionales para Chebyshev
    rp = 1
    if iir_arch == 'Chebyshev':
        rp = st.sidebar.slider("Rizado en la banda de paso (dB)", 0.1, 5.0, 1.0, 0.1)

    # Diseño del filtro IIR
    b, a = signal.iirfilter(iir_order, normal_cutoff, rp=rp, btype='low',
                             analog=False, ftype=iir_arch.lower())
    z, p, k = signal.tf2zpk(b, a)
    w, h = signal.freqz(b, a, worN=2048, fs=fs)
    w_rad = w * 2 * np.pi / fs # Frecuencia en rad/muestra para el gráfico de Bode

    # Visualización
    st.header(f"Filtro IIR ({iir_arch})")
    plot_bode(w_rad, 20 * np.log10(abs(h)), np.angle(h, deg=True), f"IIR - {iir_arch}")
    st.header("Diagrama de Polos y Ceros")
    plot_pole_zero(z, p, k, f"IIR - {iir_arch}")

st.sidebar.markdown("---")
st.sidebar.write("Desarrollado con Streamlit y SciPy.")
```
